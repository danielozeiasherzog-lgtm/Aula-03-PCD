```python
numero = int(input("Digite um número: "))

if numero <= 1:
    print("Não é primo")
elif numero == 2:
    print("É primo")
elif numero % 2 == 0:
    print("Não é primo")
else:
    primo = True

    for i in range(3, numero, 2):
        if numero % i == 0:
            primo = False
            break

    if primo:
        print("É primo")
    else:
        print("Não é primo")
```
