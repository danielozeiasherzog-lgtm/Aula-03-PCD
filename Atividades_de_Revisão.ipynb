```python
#Escreva um programa em Python que gere um número aleatório de 0 a 15, usando a biblioteca 'random' e o método 'randint()'.
#Na sequência, utilizando um loop faça com que o usuário tente acertar o número gerado.
#A cada vez que o número é acertado, armazene num arquivo o número gerado e a quantidade de tentativas que o usuário realizou para acertar.
#Quando o usuário decidir parar de adivinhar, mostre todo o histórico que ele construiu durante a execução do programa.
```

```python
# - Atividade 1 -

import random

number = random.randint(0, 15)
print(number)

numberInformed = -1
tentativas = 0

arquivo = open("historico.txt", "a")

while True:
    numberInformed = int(input("Informe um valor de 0 à 15:\n99 = Caso queira encerrar: "))

    if numberInformed == 99:
      print("Finalizou o programa")
      break

    tentativas = tentativas + 1

    if numberInformed == number:
      print("Acertou, agora foi gerado novo número!")


      arquivo.write("Número gerado: " + str(number) + " | Tentativas: " + str(tentativas) + "\n")

      number = random.randint(0, 15)
      tentativas = 0


    else:
     print("Errou!")


arquivo.close()

print("\nSalvamento")

arquivo = open("historico.txt", "r")

for linha in arquivo:
       print(linha, end="")

arquivo.close()
```

```python
#Crie um programa que permita adicionar itens para uma lista de compras.
#O usuário vai adicionando a quantidade e o valor de cada item da lista. Ao final, deve mostrar o valor total das compras e salvar a lista num arquivo .CSV.
```

```python
# - Atividade 2 -

import csv

lista = []
total = 0

while True:
    produto = input("Informe o nome do produto (99 = encerrar): ")

    if produto == "99":
        break

    quantidade = int(input("Informe a quantidade: "))
    valor = float(input("Informe o valor do produto: R$ "))

    subtotal = quantidade * valor
    total = total + subtotal

    lista.append([produto, quantidade, valor, subtotal])

print("\n--- LISTA DE COMPRAS ---")

for item in lista:
    print("Produto:", item[0])
    print("Quantidade:", item[1])
    print("Valor: R$", item[2])
    print("Subtotal: R$", item[3])
    print()

print("Valor total das compras: R$", total)


arquivo = open("lista_compras.csv", "w", newline="")

escrever = csv.writer(arquivo)

escrever.writerow(["Produto", "Quantidade", "Valor", "Subtotal"])

for item in lista:
    escrever.writerow(item)

arquivo.close()

print("\nLista salva no arquivo lista_compras.csv")
```
