```python
#Desafio 1, Estudo Independente 1 Programação para Ciência de Dados

def maiorNota(notas):
    return max(notas)


def menorNota(notas):
    return min(notas)


def calcularMedia(notas):
    return sum(notas) / len(notas)


def contarAprovados(notas):
    quantidade = 0

    for nota in notas:
        if nota >= 6:
            quantidade += 1

    return quantidade


def contarReprovados(notas):
    quantidade = 0

    for nota in notas:
        if nota < 6:
            quantidade += 1

    return quantidade


quantidadeEstudantes = int(input("Digite a quantidade de estudantes: "))

notas = []

for i in range(quantidadeEstudantes):
    nota = float(input(f"Digite a nota do estudante {i + 1}: "))
    notas.append(nota)

print(f"Maior nota: {maiorNota(notas):.2f}")
print(f"Menor nota: {menorNota(notas):.2f}")
print(f"Média da turma: {calcularMedia(notas):.2f}")
print(f"Estudantes aprovados: {contarAprovados(notas)}")
print(f"Estudantes reprovados: {contarReprovados(notas)}")

```

```python
#Desafio 3, Estudo Independente 1 Programação para Ciência de Dados

def calcularQuantidade(numeros):
    return len(numeros)


def calcularMedia(numeros):
    return sum(numeros) / len(numeros)


def calcularMediana(numeros):
    numerosOrdenados = sorted(numeros)
    quantidade = len(numerosOrdenados)

    meio = quantidade // 2

    if quantidade % 2 == 0:
        return (numerosOrdenados[meio - 1] + numerosOrdenados[meio]) / 2
    else:
        return numerosOrdenados[meio]


def calcularMaior(numeros):
    return max(numeros)


def calcularMenor(numeros):
    return min(numeros)


def contarPares(numeros):
    quantidade = 0

    for numero in numeros:
        if numero % 2 == 0:
            quantidade += 1

    return quantidade


def contarImpares(numeros):
    quantidade = 0

    for numero in numeros:
        if numero % 2 != 0:
            quantidade += 1

    return quantidade


numeros = []

print("Digite os números.")
print("Digite 0 para encerrar.")

while True:
    numero = int(input("Digite um número: "))

    if numero == 0:
        break

    numeros.append(numero)


if len(numeros) == 0:
    print("Nenhum número foi informado")
else:
    print(f"Quantidade de valores: {calcularQuantidade(numeros)}")
    print(f"Média: {calcularMedia(numeros):.2f}")
    print(f"Mediana: {calcularMediana(numeros):.2f}")
    print(f"Maior valor: {calcularMaior(numeros)}")
    print(f"Menor valor: {calcularMenor(numeros)}")
    print(f"Quantidade de pares: {contarPares(numeros)}")
    print(f"Quantidade de ímpares: {contarImpares(numeros)}")
```
